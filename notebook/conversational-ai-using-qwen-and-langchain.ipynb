# Introduction

This notebook demonstrates how to build a conversational AI chatbot using the Qwen large language model through the Hugging Face Inference API. The project leverages LangChain to simplify LLM integration and conversation management while maintaining chat history for multi-turn interactions.

The chatbot is designed to remember previous messages within a session, allowing it to generate context-aware responses and provide a more natural conversational experience. Hugging Face serves as the model provider, while LangChain handles message orchestration and memory management.

Key Features:

* Integration with Qwen via Hugging Face API
* LangChain-powered conversational workflow
* Chat message history for context retention
* Multi-turn dialogue support
* Lightweight implementation without a graphical user interface
* Fully compatible with Kaggle Notebooks

The primary objective of this project is to demonstrate how modern LLMs can be combined with LangChain memory components to create an intelligent, context-aware chatbot using a simple and scalable architecture.

# 2. Load Hugging Face Token

```python
from kaggle_secrets import UserSecretsClient

user_secrets = UserSecretsClient()
HF_TOKEN = user_secrets.get_secret("HF_TOKEN")
```

# 3. Initialize Qwen Model

```python
from langchain_huggingface import ChatHuggingFace,HuggingFaceEndpoint

llm = HuggingFaceEndpoint(
    repo_id="Qwen/Qwen2.5-7B-Instruct",
    huggingfacehub_api_token = HF_TOKEN,
    task = "text_generation",
    max_new_tokens = 512,
    temperature = 0.7,
)


chat_model = ChatHuggingFace(llm=llm)
```

# 4. Chat History Setup

```python
from langchain_core.chat_history import InMemoryChatMessageHistory

history = InMemoryChatMessageHistory()
```

# 5. Chat Function

```python
from langchain_core.messages import HumanMessage

def chat(user_input):
    history.add_user_message(user_input)
    response = chat_model.invoke(history.messages)
    history.add_ai_message(response.content)
    return response.content
```

# 6. Run Chatbot

```python
while True:
    
    question = input("You: ")

    if question.lower() in ["exit", "quit", "bye"]:
        print("Bot: Goodbye!")
        break

    answer = chat(question)

    print("\nBot:", answer)
    print("-" * 60)
```
